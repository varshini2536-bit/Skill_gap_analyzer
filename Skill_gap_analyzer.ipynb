```python

# Your skills
my_skills = ["Python", "SQL", "Machine Learning"]

# Job required skills
job_skills = ["Python", "SQL", "Excel", "Power BI"]
```

```python

match = set(my_skills) & set(job_skills)
print("Matched Skills:", match)
```

```python

missing = set(job_skills) - set(my_skills)
print("Missing Skills:", missing)
```

```python

match_percent = (len(match) / len(job_skills)) * 100
print("Match Percentage:", match_percent)
```
